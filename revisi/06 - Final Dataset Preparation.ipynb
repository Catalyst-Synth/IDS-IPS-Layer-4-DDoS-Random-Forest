```python
import pandas as pd

normal = pd.read_csv("normal_old.csv")

t1 = pd.read_csv("test1.csv")
t2 = pd.read_csv("test2.csv")
t3 = pd.read_csv("test3.csv")
t4 = pd.read_csv("test4.csv")
```

```python
drop_columns = [
    "Source_File",
    "Fwd Packets/s",
    "Total Length of Fwd Packets",
    "act_data_pkt_fwd"
]

normal = normal.drop(
    columns=drop_columns,
    errors="ignore"
)

t1 = t1.drop(columns=drop_columns, errors="ignore")
t2 = t2.drop(columns=drop_columns, errors="ignore")
t3 = t3.drop(columns=drop_columns, errors="ignore")
t4 = t4.drop(columns=drop_columns, errors="ignore")
```

```python
attack = pd.concat(
    [t1,t2,t3,t4],
    ignore_index=True
)
```

```python
attack = attack.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
attack.to_csv(
    "attack_final.csv",
    index=False
)
```

```python
dataset = pd.concat(
    [
        normal,
        attack
    ],
    ignore_index=True
)
```

```python
dataset = dataset.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
print(dataset.shape)

print()

print(dataset["Label"].value_counts())
```

```python
dataset.to_csv(
    "dataset_unbalanced_final.csv",
    index=False
)
```
