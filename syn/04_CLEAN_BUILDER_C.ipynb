```python
import os
import pandas as pd

INPUT_FOLDER = r"I:\TA\syn\lab_csv\Builder C"
OUTPUT_FOLDER = r"I:\TA\syn\lab_csv\Builder C\builder_c_clean"

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)

for file in os.listdir(INPUT_FOLDER):

    if not file.endswith(".csv"):
        continue

    path = os.path.join(
        INPUT_FOLDER,
        file
    )

    df = pd.read_csv(path)

    before = len(df)

    # =====================================
    # FILTER NOISE
    # =====================================

    df = df[
        df["Total Fwd Packets"] >= 3
    ]

    df = df[
        df["Flow Duration"] >= 0.01
    ]

    after = len(df)

    output_path = os.path.join(
        OUTPUT_FOLDER,
        file
    )

    df.to_csv(
        output_path,
        index=False
    )

    print(
        f"{file}"
        f" | BEFORE={before}"
        f" | AFTER={after}"
    )
```
