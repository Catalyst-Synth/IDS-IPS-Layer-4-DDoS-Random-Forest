```python
import pandas as pd
import os
from glob import glob
```

```python
DATASET_FOLDER = r"I:\lab_test\live traffic opi"
```

```python
csv_files = glob(os.path.join(DATASET_FOLDER, "*.csv"))

print("TOTAL FILE :", len(csv_files))
print()

for f in csv_files:
    print(os.path.basename(f))
```

```python
all_df = []

for file in csv_files:

    df = pd.read_csv(file)

    filename = os.path.basename(file).lower()

    if "syn" in filename:
        df["Label"] = "ATTACK"
    else:
        df["Label"] = "NORMAL"

    df["Source_File"] = filename

    all_df.append(df)

    print(
        f"{filename:<25}",
        f"Rows = {len(df)}"
    )
```

```python
merged_df = pd.concat(
    all_df,
    ignore_index=True
)

print()

print("MERGED SHAPE")
print(merged_df.shape)

print()

print("LABEL DISTRIBUTION")
print(
    merged_df["Label"].value_counts()
)
```

```python
print()

print("COLUMNS")

for col in merged_df.columns:
    print(col)
```

```python
print()

print("MISSING VALUES")

print(
    merged_df.isnull().sum()
)
```

```python
duplicates = merged_df.duplicated().sum()

print()

print("DUPLICATES =", duplicates)
```

```python
merged_df.to_csv(
    "opi_builder_c_live_record.csv",
    index=False
)

print()

print("Saved : opi_builder_c_live_record.csv")

print(merged_df.shape)
```
