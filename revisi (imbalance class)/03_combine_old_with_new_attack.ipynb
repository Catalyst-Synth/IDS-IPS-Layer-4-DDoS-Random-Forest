```python
import pandas as pd

df = pd.read_csv("opi_builder_c_live_record.csv")

normal_old = df[df["Label"] == "NORMAL"].copy()
attack_old = df[df["Label"] == "ATTACK"].copy()

print(normal_old.shape)
print(attack_old.shape)

normal_old.to_csv("normal_old.csv", index=False)
attack_old.to_csv("attack_old.csv", index=False)
```

```python
normal_old = pd.read_csv("normal_old.csv")
attack_new = pd.read_csv("attack_new.csv")

print(normal_old.shape)
print(attack_new.shape)
```

```python
drop_columns = [
    "Fwd Packets/s",
    "Total Length of Fwd Packets",
    "act_data_pkt_fwd"
]

normal_old = normal_old.drop(columns=drop_columns)
attack_new = attack_new.drop(columns=drop_columns)
```

```python
dataset = pd.concat(
    [
        normal_old,
        attack_new
    ],
    ignore_index=True
)

print(dataset.shape)

dataset["Label"].value_counts()
```

```python
dataset = dataset.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
dataset.to_csv(
    "dataset_unbalanced_combined.csv",
    index=False
)
```
