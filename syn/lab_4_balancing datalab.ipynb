```python
# =============================================================================
# LAB 4
# STRATIFIED BALANCING
# =============================================================================

import pandas as pd

NORMAL_FILE = r"I:\TA\syn\normal_lab_full.csv"
ATTACK_FILE = r"I:\TA\syn\attack_lab_full.csv"

normal = pd.read_csv(NORMAL_FILE)
attack = pd.read_csv(ATTACK_FILE)

print("="*80)
print("ORIGINAL")
print("="*80)

print()

print("NORMAL")
print(normal.shape)

print()

print("ATTACK")
print(attack.shape)

target_attack = len(normal)

print()
print("TARGET ATTACK :", target_attack)

# =============================================================================
# STRATIFIED SAMPLE
# =============================================================================

attack_balanced = (

    attack

    .groupby("Source_File", group_keys=False)

    .apply(

        lambda x:

        x.sample(

            n=max(

                1,

                int(
                    len(x)
                    /
                    len(attack)
                    *
                    target_attack
                )

            ),

            random_state=42

        )

    )

)

print()
print("ATTACK BALANCED")
print(attack_balanced.shape)

# =============================================================================
# MERGE
# =============================================================================

final_df = pd.concat(

    [

        normal,
        attack_balanced

    ],

    ignore_index=True

)

# =============================================================================
# SHUFFLE
# =============================================================================

final_df = (

    final_df

    .sample(

        frac=1,
        random_state=42

    )

    .reset_index(drop=True)

)

print()
print("="*80)
print("FINAL DATASET")
print("="*80)

print(final_df.shape)

print()

print(
    final_df["Label"]
    .value_counts()
)

# =============================================================================
# SAVE
# =============================================================================

OUTPUT_FILE = r"I:\TA\syn\syn_lab_balanced.csv"

final_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print()
print("SAVED :")
print(OUTPUT_FILE)

# =============================================================================
# CHECK DISTRIBUTION
# =============================================================================

print()
print("="*80)
print("SOURCE DISTRIBUTION")
print("="*80)

print(
    final_df
    .groupby(
        ["Label","Source_File"]
    )
    .size()
)
```
