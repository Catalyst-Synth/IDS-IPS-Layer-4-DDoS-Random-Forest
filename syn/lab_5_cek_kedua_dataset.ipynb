```python
# =============================================================================
# LOAD DATASET
# =============================================================================

import pandas as pd
import numpy as np

cic = pd.read_csv(
    r"I:\TA\syn\syn_clean.csv"
)

lab = pd.read_csv(
    r"I:\TA\syn\syn_lab_balanced.csv"
)

print("CIC :", cic.shape)
print("LAB :", lab.shape)
```

```python
# =============================================================================
# FEATURE LIST
# =============================================================================

FEATURES = [

    "Flow Duration",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Total Fwd Packets",
    "Total Length of Fwd Packets",
    "Fwd Packets/s",
    "SYN Flag Count",
    "ACK Flag Count",
    "Init_Win_bytes_forward",
    "act_data_pkt_fwd"

]
```

```python
# =============================================================================
# CIC STATS
# =============================================================================

print("="*80)
print("CIC DATASET")
print("="*80)

cic_stats = cic[FEATURES].describe().T

display(cic_stats)
```

```python
# =============================================================================
# LAB STATS
# =============================================================================

print("="*80)
print("LAB DATASET")
print("="*80)

lab_stats = lab[FEATURES].describe().T

display(lab_stats)
```

```python
# =============================================================================
# COMPARE TABLE
# =============================================================================

rows = []

for col in FEATURES:

    rows.append({

        "Feature": col,

        "CIC Mean":
            cic[col].mean(),

        "LAB Mean":
            lab[col].mean(),

        "CIC Median":
            cic[col].median(),

        "LAB Median":
            lab[col].median(),

        "CIC Max":
            cic[col].max(),

        "LAB Max":
            lab[col].max()

    })

compare_df = pd.DataFrame(rows)

compare_df
```

```python
# =============================================================================
# LABEL DISTRIBUTION
# =============================================================================

print("CIC")

print(
    cic["Label"]
    .value_counts()
)

print()

print("LAB")

print(
    lab["Label"]
    .value_counts()
)
```
