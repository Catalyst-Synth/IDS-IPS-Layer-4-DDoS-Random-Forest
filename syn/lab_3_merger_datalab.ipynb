```python
# =============================================================================
# LAB 3
# MERGE NORMAL + ATTACK DATASET
# =============================================================================

import pandas as pd
import glob
import os

CSV_FOLDER = r"I:\TA\syn\lab_csv"

# =============================================================================
# LOAD NORMAL FILES
# =============================================================================

normal_files = [

    f for f in glob.glob(
        os.path.join(
            CSV_FOLDER,
            "*.csv"
        )
    )

    if "normal" in os.path.basename(f).lower()

]

print("="*80)
print("NORMAL FILES")
print("="*80)

for f in normal_files:
    print(os.path.basename(f))

normal_list = []

for file in normal_files:

    df = pd.read_csv(file)

    df["Label"] = 0

    normal_list.append(df)

normal_df = pd.concat(
    normal_list,
    ignore_index=True
)

print()
print("NORMAL SHAPE")
print(normal_df.shape)

# =============================================================================
# LOAD ATTACK FILES
# =============================================================================

attack_files = [

    f for f in glob.glob(
        os.path.join(
            CSV_FOLDER,
            "*.csv"
        )
    )

    if "syn_flood" in os.path.basename(f).lower()

]

print()
print("="*80)
print("ATTACK FILES")
print("="*80)

for f in attack_files:
    print(os.path.basename(f))

attack_list = []

for file in attack_files:

    df = pd.read_csv(file)

    df["Label"] = 1

    attack_list.append(df)

attack_df = pd.concat(
    attack_list,
    ignore_index=True
)

print()
print("ATTACK SHAPE")
print(attack_df.shape)

# =============================================================================
# CLEAN
# =============================================================================

normal_df = normal_df.drop_duplicates()
attack_df = attack_df.drop_duplicates()

normal_df = normal_df.reset_index(drop=True)
attack_df = attack_df.reset_index(drop=True)

print()
print("="*80)
print("AFTER CLEAN")
print("="*80)

print("NORMAL :", normal_df.shape)
print("ATTACK :", attack_df.shape)

# =============================================================================
# SAVE
# =============================================================================

normal_output = r"I:\TA\syn\normal_lab_full.csv"
attack_output = r"I:\TA\syn\attack_lab_full.csv"

normal_df.to_csv(
    normal_output,
    index=False
)

attack_df.to_csv(
    attack_output,
    index=False
)

print()
print("="*80)
print("SAVED")
print("="*80)

print(normal_output)
print(attack_output)

# =============================================================================
# DISTRIBUTION
# =============================================================================

print()
print("="*80)
print("NORMAL DISTRIBUTION")
print("="*80)

print(
    normal_df["Source_File"]
    .value_counts()
)

print()
print("="*80)
print("ATTACK DISTRIBUTION")
print("="*80)

print(
    attack_df["Source_File"]
    .value_counts()
)
```
