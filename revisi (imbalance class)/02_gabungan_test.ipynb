```python
import pandas as pd
```

```python
df1 = pd.read_csv("test1.csv")
df2 = pd.read_csv("test2.csv")
df3 = pd.read_csv("test3.csv")
```

```python
print(df1.shape)
print(df2.shape)
print(df3.shape)
```

```python
attack_new = pd.concat(
    [df1, df2, df3],
    ignore_index=True
)
```

```python
attack_new = attack_new.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
print(attack_new.shape)

attack_new.head()
```

```python
attack_new.to_csv(
    "attack_new.csv",
    index=False
)
```
